```python
import pandas as pd
import numpy as np
import hashlib
```

```python
# TODO E2 — HASH

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan

    teks = str(nilai)
    return hashlib.sha256(
        (garam + teks).encode()
    ).hexdigest()[:12]

kolom_hash = tanda.loc[
    tanda['tindakan'] == 'hash',
    'nama_kolom'
].tolist()

for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(
            lambda x: samarkan(x, GARAM)
        )

print("Kolom yang di-hash:", kolom_hash)

if kolom_hash:
    display(aman[[col for col in kolom_hash if col in aman.columns]].head())
```

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

# Pastikan tanggal_lahir sudah bertipe datetime
df['tanggal_lahir'] = pd.to_datetime(
    df['tanggal_lahir'],
    errors='coerce'
)

# a. Hitung usia pada 1 September 2025
aman['usia'] = (
    ACUAN.year - df['tanggal_lahir'].dt.year
    - (
        (df['tanggal_lahir'].dt.month > ACUAN.month) |
        (
            (df['tanggal_lahir'].dt.month == ACUAN.month) &
            (df['tanggal_lahir'].dt.day > ACUAN.day)
        )
    ).astype(int)
)

# b. Buat kelompok umur
def buat_kelompok_umur(usia):
    if pd.isna(usia):
        return np.nan
    elif usia < 25:
        return '<25'
    elif usia <= 34:
        return '25-34'
    elif usia <= 44:
        return '35-44'
    elif usia <= 54:
        return '45-54'
    else:
        return '55+'

aman['kelompok_umur'] = aman['usia'].apply(buat_kelompok_umur)

# c. Buang tanggal_lahir dan kelurahan
aman = aman.drop(
    columns=['tanggal_lahir', 'kelurahan'],
    errors='ignore'
)

# d. Rapikan penulisan kecamatan
aman['kecamatan'] = aman['kecamatan'].astype(str).str.title()

# Hasil pengecekan
print("Jumlah kecamatan unik:", aman['kecamatan'].nunique())

print("\nJumlah peserta berdasarkan kelompok umur:")
print(aman['kelompok_umur'].value_counts())

print("\n5 data pertama:")
display(aman[['usia', 'kelompok_umur', 'kecamatan']].head())
```
