**Data:** `kunjungan_perpustakaan_2025.csv` — catatan kunjungan perpustakaan kampus, Februari–Juni 2025.
Satu baris = satu kunjungan.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/kunjungan_perpustakaan_2025.csv')
df.head()
```

---
### Tugas 1 (15 poin) — Memuat dan memeriksa

```python
# a. Cetak jumlah baris dan kolom
print(f"Jumlah baris: {df.shape[0]}, Jumlah kolom: {df.shape[1]}")
# b. Cetak jumlah nilai kosong (missing) untuk setiap kolom
print("\nJumlah missing value per kolom:")
print(df.isnull().sum())
# c. Cetak berapa mahasiswa unik (nim_hash) yang pernah berkunjung
print(f"\nJumlah mahasiswa unik: {df['nim_hash'].nunique()}")
```

---
### Tugas 2 (20 poin) — Filter

```python
# Ambil kunjungan mahasiswa prodi 'Teknik Informatika' dengan durasi_menit > 120.
fil = df[(df['prodi'] == 'Teknik Informatika') & (df['durasi_menit'] > 120)]

# Cetak jumlah barisnya dan rata-rata jumlah_pinjam pada kelompok tersebut (2 desimal).
print(f"Jumlah baris: {len(fil)}")
print(f"Rata-rata jumlah pinjam: {fil['jumlah_pinjam'].mean():.2f}")
```

---
### Tugas 3 (20 poin) — Kolom turunan

```python
# a. Buat kolom 'durasi_jam' = durasi_menit / 60, dibulatkan 2 desimal
df['durasi_jam'] = (df['durasi_menit'] / 60).round(2)

# b. Buat kolom 'lama' berisi 'Panjang' jika durasi_menit >= 90, selain itu 'Singkat'
df['lama'] = np.where(df['durasi_menit'] >= 90, 'Panjang', 'Singkat')

# c. Tampilkan 5 kunjungan dengan durasi terlama (kolom: prodi, jenis_ruang, durasi_jam)
top5 = df.sort_values(by='durasi_menit', ascending=False)[['prodi', 'jenis_ruang', 'durasi_jam']].head(5)
top5
```

---
### Tugas 4 (25 poin) — Agregasi

```python
# Buat tabel ringkasan per 'jenis_ruang' yang memuat:
#   - jumlah kunjungan
#   - rata-rata durasi_menit (1 desimal)
#   - total jumlah_pinjam
#   - rata-rata kepuasan (2 desimal)
# Urutkan dari rata-rata durasi terbesar. Simpan ke variabel 'ringkas' dan tampilkan.
```

```python
ringkasan = df.groupby('jenis_ruang').agg(
    jumlah_kunjungan=('id_kunjungan', 'count'),
    rata_rata_durasi=('durasi_menit', lambda x: round(x.mean(), 1)),
    total_jumlah_pinjam=('jumlah_pinjam', 'sum'),
    rata_rata_kepuasan=('kepuasan', lambda x: round(x.mean(), 2))
).reset_index()
ringkasan = ringkasan.sort_values(by='rata_rata_durasi', ascending=False)
ringkasan
```

---
### Tugas 5 (10 poin) — Menyimpan hasil

```python
# Simpan tabel 'ringkas' ke file 'ringkasan_ruang.csv'
# Simpan tabel 'ringkas' ke file 'ringkasan_ruang.csv'
ringkasan.to_csv('ringkasan_ruang.csv', index=False)
```
