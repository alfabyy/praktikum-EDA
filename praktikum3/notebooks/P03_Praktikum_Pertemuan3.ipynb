```python
# TODO D1 — susun tabel penandaan Anda sendiri
# Buat dictionary bernama PENANDAAN berisi pasangan:
#     'nama_kolom': ('kategori', 'tindakan')
# Tindakan yang tersedia: 'hapus', 'hash', 'generalisasi', 'pertahankan'
# Kolom yang tidak Anda daftarkan dianggap 'bukan data pribadi' + 'pertahankan'.
#
# Lalu bangun DataFrame 'tanda' dengan kolom: nama_kolom, kategori, tindakan
# dan cetak jumlah kolom per kategori.
# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom

PENANDAAN = {
    # pengenal langsung (>= 5 kolom)
    'id_peserta': ('pengenal langsung', 'hash'),
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik_dummy': ('pengenal langsung', 'hapus'),
    'no_hp': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hapus'),
    'alamat_jalan': ('pengenal langsung', 'hapus'),
    'nama_usaha': ('pengenal langsung', 'hapus'),

    # data spesifik (>= 3 kolom)
    'status_disabilitas': ('data spesifik', 'pertahankan'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'generalisasi'),
    'jumlah_anak': ('data spesifik', 'pertahankan'),

    # quasi-identifier
    'tanggal_lahir': ('quasi-identifier', 'generalisasi'),
    'jenis_kelamin': ('quasi-identifier', 'pertahankan'),
    'pendidikan_terakhir': ('quasi-identifier', 'pertahankan'),
    'kelurahan': ('quasi-identifier', 'generalisasi'),
    'kecamatan': ('quasi-identifier', 'generalisasi'),
    'kota': ('quasi-identifier', 'pertahankan'),
}

# Bangun DataFrame 'tanda'[cite: 2]
# Kolom yang tidak didaftarkan otomatis dianggap 'bukan data pribadi' + 'pertahankan'[cite: 2]
data_tanda = []
for col in df.columns:
    if col in PENANDAAN:
        kategori, tindakan = PENANDAAN[col]
    else:
        kategori, tindakan = 'bukan data pribadi', 'pertahankan'
    data_tanda.append({
        'nama_kolom': col,
        'kategori': kategori,
        'tindakan': tindakan
    })

tanda = pd.DataFrame(data_tanda)

# Cetak jumlah kolom per kategori[cite: 2]
print(tanda['kategori'].value_counts())

# Tampilkan DataFrame tanda
display(tanda)
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
# Cetak berapa kolom yang dibuang dan berapa yang tersisa.

kolom_dihapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()

aman = df.drop(columns=kolom_dihapus).copy()

jumlah_dihapus = len(kolom_dihapus)
jumlah_tersisa = aman.shape[1]

print(f"Jumlah kolom yang dibuang : {jumlah_dihapus}")
print(f"Jumlah kolom yang tersisa : {jumlah_tersisa}")
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

import hashlib

# 1. Definisikan fungsi samarkan(nilai, garam)
def samarkan(nilai, garam="rahasia_umkm_2025"):
    if pd.isna(nilai) or nilai == "" or nilai is None:
        return np.nan

    # Gabungkan garam + nilai
    teks_gabungan = str(garam) + str(nilai)

    # Hitung SHA-256 dan ambil 12 karakter pertama
    hash_object = hashlib.sha256(teks_gabungan.encode('utf-8'))
    return hash_object.hexdigest()[:12]

kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()

for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(samarkan)

print("Kolom yang di-hash:", kolom_hash)
display(aman[kolom_hash].head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

nilai_1 = "PLT-2025-1309"
nilai_2 = "PLT-2025-1512"

#nilai yang sama
hash_1a = samarkan(nilai_1)
hash_1b = samarkan(nilai_1)

print("--- a. Pembuktian Konsistensi (Nilai Sama) ---")
print(f"Nilai Asli  : {nilai_1}")
print(f"Hasil Hash 1: {hash_1a}")
print(f"Hasil Hash 2: {hash_1b}")
print(f"Apakah Hasil Identik? {hash_1a == hash_1b}\n")

#nilai berbeda
hash_2 = samarkan(nilai_2)

print("--- b. Pembuktian Keunikan (Nilai Berbeda) ---")
print(f"Nilai 1 : {nilai_1} -> Hash: {hash_1a}")
print(f"Nilai 2 : {nilai_2} -> Hash: {hash_2}")
print(f"Apakah Hasil Berbeda? {hash_1a != hash_2}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

_Hash tidak lagi tahan terhadap serangan brute-force atau dictionary attack, karena peretas dapat meng-hash semua kemungkinan kombinasi (seperti ID atau NIK) untuk mencocokkan nilai aslinya. Namun, data tersebut masih terlindungi dari dekripsi langsung secara matematis karena fungsi SHA-256 bersifat satu arah (one-way)._

```python
# TODO E4 — GENERALISASI
# a. Hitung usia peserta pada tanggal acuan 1 September 2025
# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik

ACUAN = pd.Timestamp('2025-09-01')

#Hitung usia peserta pada tanggal acuan 1 September 2025
tgl_lahir = pd.to_datetime(aman['tanggal_lahir'])
belum_ulang_tahun = (tgl_lahir.dt.month > ACUAN.month) | (
    (tgl_lahir.dt.month == ACUAN.month) & (tgl_lahir.dt.day > ACUAN.day)
)
aman['usia'] = ACUAN.year - tgl_lahir.dt.year - belum_ulang_tahun.astype(int)

#Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
bins = [-np.inf, 24, 34, 44, 54, np.inf]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(aman['usia'], bins=bins, labels=labels)

#Buang kolom tanggal_lahir dan kelurahan dari 'aman'
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'])

#Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title()

print("Distribusi kelompok umur:")
print(aman['kelompok_umur'].value_counts().sort_index())

print(f"\nJumlah kecamatan unik: {aman['kecamatan'].nunique()}")
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

def cek_k(data, kolom):
    grup_counts = data.groupby(kolom, observed=True).size()

    k_min = grup_counts.min()
    grup_1_orang = (grup_counts == 1).sum()
    total_grup = len(grup_counts)

    return pd.Series({
        'k': k_min,
        'grup_berisi_1_orang': grup_1_orang,
        'total_grup': total_grup
    })

# Menguji k-anonymity pada kolom quasi-identifier
kolom_qi = ['kecamatan', 'kelompok_umur', 'jenis_kelamin', 'pendidikan_terakhir']
hasil_cek = cek_k(aman, kolom_qi)
print(hasil_cek)
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

kolom_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']

risiko_sebelum = cek_k(df, kolom_sebelum)

print("Hasil pengukuran risiko SEBELUM penanganan:")
print(risiko_sebelum)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.

bins_f4 = [-np.inf, 34, np.inf]
labels_f4 = ['<35', '35+']
aman['kelompok_umur_f4'] = pd.cut(aman['usia'], bins=bins_f4, labels=labels_f4)

top_4_kecamatan = ['Pancoran Mas', 'Sukmajaya', 'Cimanggis', 'Beji']
aman['kecamatan_f4'] = aman['kecamatan'].apply(
    lambda x: x if x in top_4_kecamatan else 'Lainnya'
)

qi_f4 = ['kecamatan_f4', 'jenis_kelamin', 'kelompok_umur_f4']

risiko_f4 = cek_k(aman, qi_f4)

print("=== HASIL PENGUKURAN RISIKO (TODO F4) ===")
print(risiko_f4)
```
