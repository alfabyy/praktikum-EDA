**Data:** `kunjungan_perpustakaan_2025.csv` — catatan kunjungan perpustakaan kampus, Februari–Juni 2025.
Satu baris = satu kunjungan.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('kunjungan_perpustakaan_2025.csv')
df.head()
```

---
### Tugas 1 (15 poin) — Memuat dan memeriksa

```python
# a. Cetak jumlah baris dan kolom
print(df.shape)

# b. Cetak jumlah nilai kosong (missing) untuk setiap kolom
print(df.isnull().sum())

# c. Cetak berapa mahasiswa unik (nim_hash) yang pernah berkunjung
print(df["nim_hash"].nunique())

```

---
### Tugas 2 (20 poin) — Filter

```python
# Ambil kunjungan mahasiswa prodi Teknik Informatika dengan durasi_menit > 120
data = df[(df["prodi"] == "Teknik Informatika") & (df["durasi_menit"] > 120)]

# Cetak jumlah baris
print("Jumlah baris:", len(data))

# Cetak rata-rata jumlah_pinjam (2 desimal)
print("Rata-rata jumlah_pinjam:", round(data["jumlah_pinjam"].mean(), 2))
```

---
### Tugas 3 (20 poin) — Kolom turunan

```python
# a. Buat kolom durasi_jam
df["durasi_jam"] = (df["durasi_menit"] / 60).round(2)

# b. Buat kolom lama
df["lama"] = df["durasi_menit"].apply(
    lambda x: "Panjang" if x >= 90 else "Singkat"
)

# c. Tampilkan 5 kunjungan dengan durasi terlama
print(df.sort_values("durasi_menit", ascending=False)[
    ["prodi", "jenis_ruang", "durasi_jam"]
].head(5))
```

---
### Tugas 4 (25 poin) — Agregasi

```python
ringkas = df.groupby("jenis_ruang").agg(
    jumlah_kunjungan=("jenis_ruang", "size"),
    rata_rata_durasi=("durasi_menit", "mean"),
    total_jumlah_pinjam=("jumlah_pinjam", "sum"),
    rata_rata_kepuasan=("kepuasan", "mean")
)

ringkas["rata_rata_durasi"] = ringkas["rata_rata_durasi"].round(1)
ringkas["rata_rata_kepuasan"] = ringkas["rata_rata_kepuasan"].round(2)

ringkas = ringkas.sort_values("rata_rata_durasi", ascending=False)

print(ringkas)
```

---
### Tugas 5 (10 poin) — Menyimpan hasil

```python
ringkas.to_csv("ringkasan_ruang.csv")
```
