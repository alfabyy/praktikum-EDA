```python
# TODO D1 — susun tabel penandaan Anda sendiri
# Buat dictionary bernama PENANDAAN berisi pasangan:
#     'nama_kolom': ('kategori', 'tindakan')
# Tindakan yang tersedia: 'hapus', 'hash', 'generalisasi', 'pertahankan'
# Kolom yang tidak Anda daftarkan dianggap 'bukan data pribadi' + 'pertahankan'.
#
# Lalu bangun DataFrame 'tanda' dengan kolom: nama_kolom, kategori, tindakan
# dan cetak jumlah kolom per kategori.
# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom

PENANDAAN = {
    # Pengenal Langsung (Direct Identifier) >= 5 kolom
    'id_peserta': ('pengenal langsung', 'hash'),
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik_dummy': ('pengenal langsung', 'hapus'),
    'no_hp': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hapus'),
    'alamat_jalan': ('pengenal langsung', 'hapus'),

    # Data Spesifik (Sensitive Personal Data) >= 3 kolom
    'status_disabilitas': ('data spesifik', 'pertahankan'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'generalisasi'),
    'jumlah_anak': ('data spesifik', 'generalisasi'),

    # Quasi-identifier
    'kelurahan': ('quasi-identifier', 'generalisasi'),
    'kecamatan': ('quasi-identifier', 'pertahankan'),
    'kota': ('quasi-identifier', 'pertahankan'),
    'tanggal_lahir': ('quasi-identifier', 'generalisasi'),
    'jenis_kelamin': ('quasi-identifier', 'pertahankan'),
    'pendidikan_terakhir': ('quasi-identifier', 'pertahankan'),
}

# Bangun DataFrame 'tanda'
list_tanda = []
for col in df.columns:
    if col in PENANDAAN:
        kategori, tindakan = PENANDAAN[col]
    else:
        kategori, tindakan = 'bukan data pribadi', 'pertahankan'
    list_tanda.append({
        'nama_kolom': col,
        'kategori': kategori,
        'tindakan': tindakan
    })

tanda = pd.DataFrame(list_tanda)

# Cetak jumlah kolom per kategori
print("=== Jumlah Kolom per Kategori ===")
print(tanda['kategori'].value_counts())
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
# Cetak berapa kolom yang dibuang dan berapa yang tersisa.

# Mengambil daftar kolom yang bertindakan 'hapus' dari DataFrame tanda
kolom_dibuang = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()

# Membuat salinan DataFrame baru tanpa kolom-kolom tersebut
aman = df.drop(columns=kolom_dibuang).copy()

print(f"Jumlah kolom yang dibuang : {len(kolom_dibuang)} {kolom_dibuang}")
print(f"Jumlah kolom yang tersisa : {aman.shape[1]}")
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam=GARAM):
    # Kembalikan NaN jika nilainya kosong (null/None/NaN)
    if pd.isna(nilai):
        return np.nan
    # Gabungkan garam dan nilai, lalu hash dengan SHA-256 dan ambil 12 karakter pertama
    teks = str(garam) + str(nilai)
    return hashlib.sha256(teks.encode('utf-8')).hexdigest()[:12]

# Ambil kolom yang bertindakan 'hash' (misal: 'id_peserta')
kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()

# Terapkan fungsi samarkan ke kolom tersebut pada DataFrame aman
for kol in kolom_hash:
    aman[kol] = aman[kol].apply(lambda x: samarkan(x, GARAM))

# Tampilkan 5 baris pertama hasil hash
print(f"Kolom yang disamarkan (hash): {kolom_hash}")
aman[kolom_hash].head()
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca

nilai_1 = "PLT-2025-1309"
nilai_2 = "PLT-2025-1310"

# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik

hasil_a1 = samarkan(nilai_1, GARAM)
hasil_a2 = samarkan(nilai_1, GARAM)

print("=== a. Uji Nilai Sama (Konsistensi Determinisme) ===")
print(f"Hash 1 : {hasil_a1}")
print(f"Hash 2 : {hasil_a2}")
print(f"Apakah identik? -> {hasil_a1 == hasil_a2}\n")

# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

hasil_b = samarkan(nilai_2, GARAM)

print("=== b. Uji Nilai Berbeda (Collision-Resistant / Keunikan) ===")
print(f"Hash nilai_1 ({nilai_1}): {hasil_a1}")
print(f"Hash nilai_2 ({nilai_2}): {hasil_b}")
print(f"Apakah berbeda? -> {hasil_a1 != hasil_b}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

Yang masih terlindungi: Data yang sudah dibuang permanen (seperti NIK dummy, no HP, alamat, email) tetap aman dan tidak dapat dipulihkan.

Yang tidak terlindungi: Keamanan proses hashing runtuh terhadap HACK. Siapapun yang memiliki dataset dan nilai GARAM dapat mencocokkan atau membuat kamus nilai ID asli secara mudah untuk memetakan kembali identitas asli peserta.

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta pada tanggal acuan 1 September 2025

aman['usia'] = (ACUAN.year - aman['tanggal_lahir'].dt.year) - (
    (ACUAN.month < aman['tanggal_lahir'].dt.month) |
    ((ACUAN.month == aman['tanggal_lahir'].dt.month) & (ACUAN.day < aman['tanggal_lahir'].dt.day))
)

# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+

batas_umur = [-float('inf'), 24, 34, 44, 54, float('inf')]
label_umur = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(aman['usia'], bins=batas_umur, labels=label_umur)

# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'

aman = aman.drop(columns=['tanggal_lahir', 'kelurahan', 'usia'])

# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL

aman['kecamatan'] = aman['kecamatan'].str.strip().str.title()

# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik

print("=== Distribusi Kelompok Umur ===")
print(aman['kelompok_umur'].value_counts())

print(f"\nJumlah nilai unik kecamatan: {aman['kecamatan'].nunique()}")
print("Daftar kecamatan:", aman['kecamatan'].unique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

def cek_k(data, kolom):
    # Hitung jumlah anggota untuk setiap kombinasi nilai pada kolom quasi-identifier
    ukuran_grup = data.groupby(kolom, observed=False).size()

    # Ambil hanya grup yang memiliki data (ukuran > 0)
    ukuran_grup = ukuran_grup[ukuran_grup > 0]

    # 1. k = ukuran grup terkecil (minimum group size)
    k = ukuran_grup.min()

    # 2. Banyaknya kombinasi grup yang hanya berisi tepat 1 orang (sangat berisiko teridentifikasi)
    grup_1_orang = (ukuran_grup == 1).sum()

    # 3. Total seluruh grup kombinasi unik yang terbentuk
    total_grup = len(ukuran_grup)

    return pd.Series({
        'k': k,
        'grup_1_orang': grup_1_orang,
        'total_grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

kolom_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']

hasil_f2 = cek_k(df, kolom_sebelum)
print(hasil_f2)
```

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1

kolom_sesudah = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']

hasil_f3 = cek_k(aman, kolom_sesudah)
print(hasil_f3)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.

map_pendidikan = {
    'SD': '< SMA',
    'SMP': '< SMA',
    'SMA/SMK': 'SMA/SMK',
    'D3': 'Perguruan Tinggi',
    'S1': 'Perguruan Tinggi',
    'S2': 'Perguruan Tinggi'
}
aman['pendidikan_gen'] = aman['pendidikan_terakhir'].map(map_pendidikan)

# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',

map_umur = {
    '<25': '<35',
    '25-34': '<35',
    '35-44': '35-44',
    '45-54': '45+',
    '55+': '45+'
}
aman['kelompok_umur_gen'] = aman['kelompok_umur'].map(map_umur)

# atau mengeluarkan satu kolom dari daftar quasi-identifier.

qi_final = ['jenis_kelamin', 'kelompok_umur_gen', 'pendidikan_gen']

# Catat setiap pilihan dan dampaknya terhadap k.

hasil_f4 = cek_k(aman, qi_final)
print("=== Hasil Pengukuran Risiko (F4) ===")
print(hasil_f4)

print("\n=== Ukuran Seluruh Grup Kombinasi (Terurut) ===")
grup_final = aman.groupby(qi_final, observed=False).size()
print(grup_final.sort_values())

# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.
```

  - pendidikan_gen digabungkan menjadi 3 jenjang: < SMA (SD & SMP), SMA/SMK, dan Perguruan Tinggi (D3, S1, S2).   
  - Kolom kecamatan dikeluarkan dari daftar quasi-identifier.

- Nilai $k$ yang dicapai:$k = 11$ (sudah melampaui target $k \ge 5$), dengan jumlah kelompok berukuran satu orang sebanyak 0 grup.

- Informasi yang dikorbankan (information loss):
  1. Granularitas Geografis: Analisis perbandingan performa atau sebaran UMKM antar-kecamatan di Kota Depok hilang karena kolom kecamatan tidak lagi dipakai dalam pelaporan kombinasi publik.   
  2. Detail Tingkat Pendidikan: Hilangnya kemampuan membedakan lulusan perguruan tinggi (misal D3 vs S1 vs S2) dan sekolah dasar (SD vs SMP) karena telah dilebur ke kategori agregat.   
  3. Presisi Usia: Rentang usia menjadi lebih lebar/kasar (hanya 3 kategori besar), sehingga variasi peserta sangat muda (<25 tahun) atau usia pensiun (55+ tahun) tidak dapat dianalisis secara spesifik lagi.

```python
# TODO F5 — sel kecil pada tabel agregat
# a. Buat tabel silang jumlah peserta menurut kecamatan x kategori_usaha

tabel_silang = pd.crosstab(aman['kecamatan'], aman['kategori_usaha'])
print("=== Tabel Silang Kecamatan x Kategori Usaha ===")
display(tabel_silang)

# b. Hitung berapa sel yang terisi tetapi berisi kurang dari 5 peserta

mask_sel_kecil = (tabel_silang > 0) & (tabel_silang < 5)
jumlah_sel_kecil = mask_sel_kecil.sum().sum()

print(f"\nJumlah sel yang terisi tetapi berisi kurang dari 5 peserta: {jumlah_sel_kecil} sel")

# Expected: ada sel bernilai 1-4; sel seperti ini tidak boleh ditampilkan apa adanya di laporan
```
